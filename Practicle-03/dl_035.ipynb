```python
# ============================================================
# PRACTICAL 3
# Forward Propagation and Backpropagation using TensorFlow/Keras
# Effect of Learning Rate and Number of Epochs
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras

from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay


# ============================================================
# 2. LOAD MNIST DATASET
# ============================================================

(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)


# ============================================================
# 3. NORMALIZE THE DATA
# ============================================================

X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

print("\nPixel value range after normalization:")
print("Minimum:", X_train.min())
print("Maximum:", X_train.max())


# ============================================================
# 4. VISUALIZE SAMPLE IMAGES
# ============================================================

plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 5. CREATE MLP MODEL
# ============================================================

def create_model(learning_rate):

    model = keras.Sequential([
        keras.layers.Flatten(input_shape=(28, 28)),

        keras.layers.Dense(128, activation="relu"),

        keras.layers.Dense(64, activation="relu"),

        keras.layers.Dense(10, activation="softmax")
    ])

    optimizer = keras.optimizers.Adam(
        learning_rate=learning_rate
    )

    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# 6. EXPERIMENT 1 — DIFFERENT LEARNING RATES
# ============================================================

learning_rates = [0.001, 0.01, 0.1]

learning_rate_results = {}

for lr in learning_rates:

    print("\n====================================")
    print("Learning Rate:", lr)
    print("====================================")

    model = create_model(lr)

    history = model.fit(
        X_train,
        y_train,
        epochs=10,
        batch_size=128,
        validation_split=0.1,
        verbose=1
    )

    test_loss, test_accuracy = model.evaluate(
        X_test,
        y_test,
        verbose=0
    )

    learning_rate_results[lr] = {
        "model": model,
        "history": history,
        "test_loss": test_loss,
        "test_accuracy": test_accuracy
    }

    print("Test Accuracy:", test_accuracy)
    print("Test Accuracy (%):", test_accuracy * 100)


# ============================================================
# 7. COMPARE LEARNING RATES
# ============================================================

print("\nLearning Rate Results")
print("--------------------------------")

for lr, result in learning_rate_results.items():

    print(
        f"Learning Rate: {lr} | "
        f"Test Accuracy: {result['test_accuracy'] * 100:.2f}% | "
        f"Test Loss: {result['test_loss']:.4f}"
    )


# ============================================================
# 8. PLOT EFFECT OF LEARNING RATE
# ============================================================

lr_values = []
lr_accuracies = []

for lr, result in learning_rate_results.items():

    lr_values.append(lr)
    lr_accuracies.append(result["test_accuracy"])

plt.figure(figsize=(8, 5))

plt.plot(
    lr_values,
    lr_accuracies,
    marker="o"
)

plt.xlabel("Learning Rate")
plt.ylabel("Test Accuracy")
plt.title("Effect of Learning Rate on Test Accuracy")
plt.xscale("log")
plt.grid(True)
plt.show()


# ============================================================
# 9. EXPERIMENT 2 — DIFFERENT NUMBER OF EPOCHS
# ============================================================

epoch_values = [5, 10, 20, 30]

epoch_results = {}

for epochs in epoch_values:

    print("\n====================================")
    print("Number of Epochs:", epochs)
    print("====================================")

    model = create_model(0.001)

    history = model.fit(
        X_train,
        y_train,
        epochs=epochs,
        batch_size=128,
        validation_split=0.1,
        verbose=1
    )

    test_loss, test_accuracy = model.evaluate(
        X_test,
        y_test,
        verbose=0
    )

    epoch_results[epochs] = {
        "model": model,
        "history": history,
        "test_loss": test_loss,
        "test_accuracy": test_accuracy
    }

    print("Test Accuracy:", test_accuracy)
    print("Test Accuracy (%):", test_accuracy * 100)


# ============================================================
# 10. COMPARE NUMBER OF EPOCHS
# ============================================================

print("\nEpoch Results")
print("--------------------------------")

for epochs, result in epoch_results.items():

    print(
        f"Epochs: {epochs} | "
        f"Test Accuracy: {result['test_accuracy'] * 100:.2f}% | "
        f"Test Loss: {result['test_loss']:.4f}"
    )


# ============================================================
# 11. PLOT EFFECT OF EPOCHS
# ============================================================

epoch_values_plot = []
epoch_accuracies = []

for epochs, result in epoch_results.items():

    epoch_values_plot.append(epochs)
    epoch_accuracies.append(result["test_accuracy"])

plt.figure(figsize=(8, 5))

plt.plot(
    epoch_values_plot,
    epoch_accuracies,
    marker="o"
)

plt.xlabel("Number of Epochs")
plt.ylabel("Test Accuracy")
plt.title("Effect of Number of Epochs on Test Accuracy")
plt.grid(True)
plt.show()


# ============================================================
# 12. TRAIN FINAL MODEL
# ============================================================

final_model = create_model(0.001)

final_history = final_model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)


# ============================================================
# 13. FINAL MODEL EVALUATION
# ============================================================

final_loss, final_accuracy = final_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\nFinal Model Performance")
print("--------------------------------")
print("Test Loss:", final_loss)
print("Test Accuracy:", final_accuracy)
print("Test Accuracy (%):", final_accuracy * 100)


# ============================================================
# 14. FORWARD PROPAGATION / PREDICTIONS
# ============================================================

y_pred_prob = final_model.predict(X_test)

y_pred = np.argmax(
    y_pred_prob,
    axis=1
)

print("\nFirst 10 Predictions:")
print(y_pred[:10])

print("\nFirst 10 Actual Labels:")
print(y_test[:10])


# ============================================================
# 15. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=np.arange(10)
)

disp.plot()

plt.title("MNIST Confusion Matrix")
plt.show()


# ============================================================
# 16. TRAINING ACCURACY AND VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    final_history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    final_history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training and Validation Accuracy")

plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# 17. TRAINING LOSS AND VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    final_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    final_history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")

plt.legend()
plt.grid(True)
plt.show()
```
