```python

# ============================================================
# PRACTICAL 6: CNN FOR TOMATO LEAF DISEASE CLASSIFICATION
# Dataset: PlantVillage Tomato Disease Dataset
# Task: Multi-Class Image Classification
# Platform: Google Colab
# ============================================================


# 1. IMPORT REQUIRED LIBRARIES

import os
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)

import seaborn as sns

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout,
    Rescaling
)

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping


# Set random seed
tf.random.set_seed(42)
np.random.seed(42)


# 2. DOWNLOAD TOMATO DATASET

# Download PlantVillage dataset from GitHub
dataset_url = (
    "https://github.com/spMohanty/PlantVillage-Dataset/"
    "archive/refs/heads/master.zip"
)

zip_path = tf.keras.utils.get_file(
    fname="PlantVillage.zip",
    origin=dataset_url,
    extract=False
)

# Extract downloaded dataset
extract_path = "/content/PlantVillage"

if not os.path.exists(extract_path):

    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall("/content/")

print("Dataset downloaded and extracted successfully!")


# 3. SELECT TOMATO IMAGES

# Locate the PlantVillage color images directory
base_path = Path(
    "/content/PlantVillage-Dataset-master/raw/color"
)

# Select only folders related to tomato diseases
tomato_folders = sorted([
    folder for folder in base_path.iterdir()
    if folder.is_dir() and folder.name.startswith("Tomato___")
])

print("\nTomato Disease Classes:")

for folder in tomato_folders:
    print(folder.name)


# 4. PREPARE IMAGE DATASET

# Image size and batch size
IMG_SIZE = 128
BATCH_SIZE = 32

image_paths = []
labels = []

# Collect image paths and corresponding labels
for folder in tomato_folders:

    class_name = folder.name

    for image_file in folder.glob("*"):

        if image_file.suffix.lower() in [
            ".jpg", ".jpeg", ".png"
        ]:

            image_paths.append(str(image_file))
            labels.append(class_name)

# Create DataFrame
df = pd.DataFrame({
    "image_path": image_paths,
    "label": labels
})

print("\nTotal Images:", len(df))
print("\nClass Distribution:")
print(df["label"].value_counts())


# 5. SPLIT DATASET

# Split into 80% training, 10% validation, 10% testing
train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["label"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["label"]
)

print("\nTraining Images:", len(train_df))
print("Validation Images:", len(val_df))
print("Testing Images:", len(test_df))


# 6. LOAD AND PREPROCESS IMAGES

# Data augmentation for training images
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True
)

# Validation and testing data
test_datagen = ImageDataGenerator(
    rescale=1./255
)

# Training generator
train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_df,
    x_col="image_path",
    y_col="label",
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=True,
    seed=42
)

# Validation generator
val_generator = test_datagen.flow_from_dataframe(
    dataframe=val_df,
    x_col="image_path",
    y_col="label",
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

# Testing generator
test_generator = test_datagen.flow_from_dataframe(
    dataframe=test_df,
    x_col="image_path",
    y_col="label",
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)


# 7. DISPLAY SAMPLE IMAGES

images, image_labels = next(train_generator)

plt.figure(figsize=(12, 8))

for i in range(min(9, len(images))):

    plt.subplot(3, 3, i + 1)

    plt.imshow(images[i])

    class_index = np.argmax(image_labels[i])

    class_name = list(
        train_generator.class_indices.keys()
    )[class_index]

    plt.title(class_name)
    plt.axis("off")

plt.tight_layout()
plt.show()


# 8. BUILD CNN MODEL

num_classes = len(train_generator.class_indices)

model = Sequential([

    # First convolution block
    Conv2D(
        32,
        (3, 3),
        activation='relu',
        input_shape=(IMG_SIZE, IMG_SIZE, 3)
    ),

    MaxPooling2D((2, 2)),

    # Second convolution block
    Conv2D(
        64,
        (3, 3),
        activation='relu'
    ),

    MaxPooling2D((2, 2)),

    # Third convolution block
    Conv2D(
        128,
        (3, 3),
        activation='relu'
    ),

    MaxPooling2D((2, 2)),

    # Convert feature maps into a single vector
    Flatten(),

    # Fully connected layer
    Dense(128, activation='relu'),

    # Dropout to reduce overfitting
    Dropout(0.5),

    # Output layer for disease classification
    Dense(num_classes, activation='softmax')
])


# 9. COMPILE CNN MODEL

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Display model architecture
model.summary()


# 10. TRAIN CNN MODEL

# Stop training if validation loss does not improve
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=15,
    callbacks=[early_stop]
)

print("\nCNN Model Training Completed!")


# 11. VISUALIZE TRAINING ACCURACY

plt.figure(figsize=(10, 5))

plt.plot(
    history.history['accuracy'],
    label='Training Accuracy'
)

plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)

plt.title("CNN Training and Validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)
plt.show()


# 12. VISUALIZE TRAINING LOSS

plt.figure(figsize=(10, 5))

plt.plot(
    history.history['loss'],
    label='Training Loss'
)

plt.plot(
    history.history['val_loss'],
    label='Validation Loss'
)

plt.title("CNN Training and Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)
plt.show()


# 13. EVALUATE MODEL ON TEST DATA

test_loss, test_accuracy = model.evaluate(
    test_generator
)

print("\n========== MODEL EVALUATION ==========")

print("Test Loss:", round(test_loss, 4))
print("Test Accuracy:", round(test_accuracy * 100, 2), "%")


# 14. GENERATE PREDICTIONS

# Predict disease classes
test_generator.reset()

predictions = model.predict(test_generator)

predicted_classes = np.argmax(
    predictions,
    axis=1
)

actual_classes = test_generator.classes

class_names = list(
    test_generator.class_indices.keys()
)

# Calculate accuracy
accuracy = accuracy_score(
    actual_classes,
    predicted_classes
)

print("\nClassification Accuracy:", round(accuracy, 4))


# 15. CLASSIFICATION REPORT

print("\n========== CLASSIFICATION REPORT ==========")

print(
    classification_report(
        actual_classes,
        predicted_classes,
        target_names=class_names,
        zero_division=0
    )
)


# 16. CONFUSION MATRIX

cm = confusion_matrix(
    actual_classes,
    predicted_classes
)

plt.figure(figsize=(12, 10))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("CNN Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.xticks(rotation=90)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()


# 17. PREDICT A SAMPLE LEAF IMAGE

# Get one image from test dataset
sample_images, sample_labels = next(test_generator)

# Predict sample image
sample_prediction = model.predict(
    sample_images[:1]
)

predicted_index = np.argmax(
    sample_prediction[0]
)

actual_index = np.argmax(
    sample_labels[0]
)

print("\n========== SAMPLE PREDICTION ==========")

print("Actual Disease:", class_names[actual_index])

print("Predicted Disease:", class_names[predicted_index])

# Display sample image
plt.figure(figsize=(5, 5))

plt.imshow(sample_images[0])

plt.title(
    "Predicted: " + class_names[predicted_index]
)

plt.axis("off")
plt.show()


# 18. SAVE TRAINED MODEL

model.save("tomato_cnn_model.keras")

print("\nTrained CNN model saved successfully!")


# 19. SAVE CLASSIFICATION RESULTS

results_df = pd.DataFrame({
    "Actual Disease": [
        class_names[i] for i in actual_classes
    ],
    "Predicted Disease": [
        class_names[i] for i in predicted_classes
    ]
})

results_df.to_csv(
    "cnn_prediction_results.csv",
    index=False
)

print("\nPrediction results saved successfully!")


# 20. SAVE CLASS NAMES

class_mapping = pd.DataFrame({
    "Class Index": list(range(len(class_names))),
    "Disease Name": class_names
})

class_mapping.to_csv(
    "class_names.csv",
    index=False
)


# 21. DOWNLOAD OUTPUT FILES

from google.colab import files

files.download("tomato_cnn_model.keras")

files.download("cnn_prediction_results.csv")

files.download("class_names.csv")

print("\nPRACTICAL COMPLETED SUCCESSFULLY!")
```
